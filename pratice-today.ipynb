### C7. frozenset + shallow vs deep copy
Write both copy examples with nested list


```python
import copy

print(frozenset([1, 2, 2, 3]))

original = [[10, 20], [30, 40]]
shallow = copy.copy(original)
deep = copy.deepcopy(original)
shallow[0][0] = 100
print("after shallow change:", original)  # changed
deep[0][0] = 999
print("original still:", original)
print("deep:", deep)
```
