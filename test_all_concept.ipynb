```python
class Bankaccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance
    def deposit(self,amount):
        self.balance=self.balance+amount
    def withdraw(self,amount):
        if amount<= self.balance:
            self.balance=self.balance- amount
            print("withdrawal sucesfull")
        else:
            print("insufficient funds")
    def display(self):
        print("account_holder:",self.account_holder)
        print("balance:",self.balance)
account1=Bankaccount("sid",10000)
account1.deposit(5000)
account1.withdraw(2000)
account1.display()
```

```python
#assert : The assert statement is used to check whether a condition is true. If the condition is false, Python raises an AssertionError
file_size=0
try:
    assert file_size>0,"file is empty"
    print("file contains data")
except AssertionError as e:
    print("error",e)
    
```

```python
#Custom Exceptions :A custom exception is an exception that we create ourselves to handle specific errors in our programs.
#We create custom exceptions using a class that inherits from Python's Exception class.
class EmptyFileError(Exception):
    pass
try:
    file_size = 0

    if file_size == 0:
        raise EmptyFileError("File is empty")

except EmptyFileError as e:
    print("Error:", e)
```

```python
#flush Normally, Python may temporarily keep written data in memory (buffer) before sending it to the file.

#flush() tells Python to send that pending data to the file immediately.
#flush() is useful when you want to make sure data is written to the file before the file is closed.
```
