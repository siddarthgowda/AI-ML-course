## 9. Conditional statements

## 10. `for` loop

**Definition:** A `for` loop repeats a block a fixed number of times. `range(stop)` produces numbers from `0` up to, and not including, `stop`.

`i` is the counter. Any name works: `i`, `n`, `count`.

- `range(start, stop)` begins at `start` and stops before `stop`.
- `range(start, stop, step)` jumps by `step`.
- A negative step counts backward.


````python
print("--- good morning, 5 times ---")
for i in range(5):
    print("good morning")

print("--- range(5) ---")
for i in range(5):
    print(i)

print("--- 1 to 10 ---")
for i in range(1, 11):
    print(i)

print("--- step of 2 ---")
for i in range(1, 11, 2):
    print(i)

print("--- 10 down to 1 ---")
for i in range(10, 0, -1):
    print(i)

print("--- table of 7 ---")
for i in range(1, 11):
    print(i * 7)
````

Sum from 1 to 10. `s` starts at `0` and grows by `i` on every pass. `print` sits outside the loop, so only the final total is shown.


````python
s = 0
for i in range(1, 11):
    s = s + i
print(s)
````

## 11. `while` loop

**Definition:** A `while` loop repeats as long as its condition stays `True`. The body must change something in that condition. Otherwise the loop never ends.

Each pass prints `a`, then subtracts 1. When `a` becomes `-1`, `a >= 0` is false and the loop stops.


````python
a = 5
while a >= 0:
    print(a)
    a = a - 1
````

## 12. `break` and `continue`

**Definition:**

- `break` leaves the loop immediately.
- `continue` skips the rest of the current pass and starts the next one.


````python
print("--- break at 6 ---")
for i in range(1, 11):
    if i == 6:
        break
    print(i)

print("--- continue skips 6 ---")
for i in range(1, 11):
    if i == 6:
        continue
    print(i)
````

Replace 5 with the word `FIVE`.


````python
for i in range(1, 11):
    if i == 5:
        print("FIVE")
    else:
        print(i)
````

Numbers from 1 to 100 that are divisible by 7. The second loop uses a step and prints the same sequence.


````python
print("--- remainder check ---")
for i in range(1, 101):
    if i % 7 == 0:
        print(i)

print("--- step of 7 ---")
for i in range(7, 101, 7):
    print(i)
````

## 1. Strings

**Indexing.** Positions start at `0` from the left and `-1` from the right.


````python
name = "Python"
print(name[0])
print(name[1])
print(name[-1])
print(name[-2])
````

````python
name = "Python"
print(name[0:3])
print(name[:5])
print(name[2:])

my_string = "have a nice day"
print(my_string[7:11])
print(my_string[::2])
print(my_string[::-1])
````

## 1. Dictionary

**Definition:** A dictionary stores data as key-value pairs inside curly braces. The key is the label. The value is the data stored under that label.

```python
employee = {
    "name": "Rahul",
    "age": 24,
    "department": "AI",
    "salary": 1000
}
```

Read one value with `dictionary[key]`.

`employee["email"]` raises `KeyError` when that key was never stored.

A key appears once. If the same key is written twice, the later value replaces the earlier one.

Keys in this class are strings or numbers. Values can be any type. A list cannot be a key.


````python
employee = {
    "name": "Rahul",
    "age": 24,
    "department": "AI",
    "salary": 1000
}
print(employee)
print(employee["name"])
print(employee["age"])

try:
    print(employee["email"])
except KeyError as error:
    print("KeyError:", error)

repeat = {"name": "Rahul", "name": "Priya"}
print(repeat)

codes = {1: "Python", 2: "SQL"}
print(codes[1])
````

## 2. Add, update, and remove

**Add** a pair by assigning a new key.

**Update** a value by assigning an existing key.

**Remove** one pair with `pop` or `del`.

- `pop(key)` removes the pair and returns the value.
- `del dictionary[key]` removes the pair and returns nothing.
- `del dictionary` deletes the whole name.

A new key is added at the end. A dictionary has no index, so there is no `insert` at a chosen position.

You cannot rename a key in place. Remove the old key, then add the new one.

The dictionary itself can change: add, update, and delete. The class rule was: keys stay as they were written, values can be replaced.


````python
employee = {
    "name": "Rahul",
    "age": 24,
    "department": "AI",
    "salary": 1000
}

employee["city"] = "Mumbai"
print(employee)

employee["salary"] = 2000
print(employee["salary"])

removed = employee.pop("city")
print(removed)
print(employee)

del employee["age"]
print(employee)

employee[" "] = "space key"
print(employee)
````

| `keys()` | All keys |
| `values()` | All values |
| `items()` | Each pair as `(key, value)` |
| `get(key, default)` | The value, or `default` when the key is missing |

## Day 4. Functions

**Definition:** A function is a reusable block of code designed to perform a specific task.

Instead of writing the same code multiple times, we can write it once inside a function and call it whenever required.

Why we use functions:

- Avoid repeating code.
- Reuse code whenever required.
- Make programs easier to read.
- Divide a large task into smaller tasks.
- Make code easier to test and debug.

### Types of functions

| Type | Definition | Example |
|---|---|---|
| Built-in | Already provided by Python | `print()`, `len()`, `sum()`, `max()`, `min()`, `type()` |
| User-defined | Created by you with `def` | `def greet():` |
| Lambda | Small unnamed function with `lambda` | `lambda x: x * x` |
| Recursive | A function that calls itself. Needs a base case to stop | `countdown(n)` |

### Function syntax

Use `def`, a name, parentheses, and a colon. Indent the body, commonly four spaces. Wrong indentation causes `IndentationError`.


````python
def greet():
    print("Hello, how are you?")

greet()
greet()
````

````python
print(len([10, 20, 30]))
print(sum([10, 20, 30]))

def multiply(a, b):
    return a * b

print(multiply(4, 5))

square = lambda x: x * x
print(square(5))

def countdown(number):
    if number == 0:
        print("Done")
    else:
        print(number)
        countdown(number - 1)

countdown(3)
````

## Day 5. *args, **kwargs, lambda, comprehensions


### 1. Variable number of arguments (`*args`)

**Definition:** `*args` allows a function to accept any number of positional arguments without specifying the exact number of arguments in advance.

Python collects all the positional arguments passed to `*args` and stores them inside a tuple.

Syntax: `def function_name(*args):`

**Important points:**
- `*args` accepts multiple positional arguments.
- Arguments are stored in a tuple.
- The name `args` is not compulsory. The `*` symbol is important.
- You can use a different name, such as `*numbers` or `*values`.


````python
def display(*args):
    print(args)

display(10, 20, 30, 40)

def calculate_sum(*numbers):
    total = sum(numbers)
    print("Total:", total)

calculate_sum(10, 20, 30)
calculate_sum(5, 10, 15, 20)
````

### 2. Variable keyword arguments (`**kwargs`)

**Definition:** `**kwargs` allows a function to accept any number of keyword arguments without specifying the exact number of arguments in advance.

Python collects all the keyword arguments and stores them inside a dictionary in the form of key-value pairs.

Syntax: `def function_name(**kwargs):`

**Important points:**
- `**kwargs` accepts multiple keyword arguments.
- Arguments are stored in a dictionary.
- Each argument has a key and a value.
- The name `kwargs` is not compulsory. The `**` symbol is important.


````python
def student_details(**kwargs):
    print(kwargs)

student_details(name="Rahul", age=22, course="Python")

def employee(**details):
    for key, value in details.items():
        print(key, ":", value)

employee(name="Rahul", salary=30000, department="IT")
````

### 3. Combining `*args` and `**kwargs`

**Definition:** We can use `*args` and `**kwargs` together in a function when we want to accept both multiple positional arguments and multiple keyword arguments.

`*args` collects positional arguments into a tuple. `**kwargs` collects keyword arguments into a dictionary.

Syntax: `def function_name(*args, **kwargs):`

**Important points:**
- `*args` must come before `**kwargs` in the function definition.
- Positional arguments are collected in a tuple.
- Keyword arguments are collected in a dictionary.
- You can use both in the same function.


````python
def student(*subjects, **details):
    print("Subjects:", subjects)
    print("Details:", details)

student("Python", "SQL", name="Rahul", age=22)
````

### 4. Lambda function

**Definition:** A lambda function is a small anonymous function created using the `lambda` keyword.

It can accept any number of arguments but contains only one expression. The expression is evaluated and its result is automatically returned.

Syntax: `lambda arguments: expression`

**Important points:**
- Lambda functions are anonymous functions.
- They are created using the `lambda` keyword.
- They can accept multiple arguments.
- They contain only one expression.
- The result of the expression is returned automatically.
- They are commonly used with `map()`, `filter()`, and `sorted()`.


````python
square = lambda x: x * x
print(square(5))

add = lambda a, b: a + b
print(add(10, 20))

subtract = lambda a, b: a - b
print(subtract(20, 10))

add_ten = lambda x: x + 10
print(add_ten(5))
````

### 5. List comprehension

**Definition:** List comprehension is a shorter and more readable way to create a new list from an existing iterable.

It allows us to apply an expression to each element and optionally include a condition to select specific elements.

Syntax: `new_list = [expression for item in iterable]`

**Important points:**
- List comprehension creates a new list.
- It can apply an expression to every element.
- It can include conditions using `if`.
- It can use if-else to choose between two expressions.
- It makes code shorter than writing a normal for loop.


````python
numbers = [1, 2, 3, 4, 5]
squares = [i * i for i in numbers]
print(squares)

new_list = [i + 10 for i in numbers]
print(new_list)

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = [i for i in numbers if i % 2 == 0]
print(even_numbers)

result = [i + 10 if i > 3 else i for i in numbers]
print(result)
````

### 6. Set comprehension

**Definition:** Set comprehension is a shorter way to create a set from an existing iterable.

It allows us to apply an expression to each element and optionally use conditions to select elements.

Syntax: `new_set = {expression for item in iterable}`

**Important points:**
- Set comprehension creates a new set.
- Curly brackets `{ }` are used.
- Sets do not store duplicate values.
- Sets are unordered collections.
- Conditions can be used to filter elements.


````python
numbers = [1, 2, 3, 4, 5]
squares = {i * i for i in numbers}
print(squares)

numbers = [1, 2, 2, 3, 3, 4, 5]
unique_numbers = {i for i in numbers}
print(unique_numbers)

numbers = [1, 2, 3, 4, 5, 6]
even_numbers = {i for i in numbers if i % 2 == 0}
print(even_numbers)
````

### 7. Dictionary comprehension

**Definition:** Dictionary comprehension is a shorter way to create a dictionary from an existing iterable.

It allows us to create key-value pairs using an expression and optionally apply conditions.

Syntax: `new_dict = {key: value for item in iterable}`

**Important points:**
- Dictionary comprehension creates a dictionary.
- Curly brackets `{ }` are used.
- Key-value pairs are separated using a colon `:`.
- Each key must be unique.
- Conditions can be used to filter elements.


````python
numbers = [1, 2, 3, 4, 5]
squares = {i: i * i for i in numbers}
print(squares)

names = ["Rahul", "Amit", "Priya"]
name_lengths = {name: len(name) for name in names}
print(name_lengths)

numbers = [1, 2, 3, 4, 5, 6]
even_squares = {i: i * i for i in numbers if i % 2 == 0}
print(even_squares)
````

### 8. map()

**Definition:** The `map()` function applies a given function to every element of an iterable and returns a map iterator containing the results.

It is mainly used when we want to transform or modify every element in a collection.

Syntax: `map(function, iterable)`

**Important points:**
- `map()` applies a function to every element.
- It is used for transforming data.
- It returns a map iterator, not a list.
- Use `list()` to convert the result into a list.
- It does not modify the original iterable.


````python
numbers = [1, 2, 3, 4, 5]
result = list(map(lambda x: x * 2, numbers))
print(result)

numbers = [10, 20, 30, 40]
result = list(map(lambda x: x + 10, numbers))
print(result)

names = ["rahul", "amit", "priya"]
result = list(map(lambda name: name.upper(), names))
print(result)
````

### 9. filter()

**Definition:** The `filter()` function selects elements from an iterable based on a condition.

It applies a function to each element and keeps only those elements for which the function returns `True`.

Syntax: `filter(function, iterable)`

**Important points:**
- `filter()` selects elements based on a condition.
- It keeps elements for which the function returns `True`.
- It returns a filter iterator, not a list.
- Use `list()` to convert the result into a list.
- It does not modify the original iterable.


````python
numbers = [1, 2, 3, 4, 5, 6]
result = list(filter(lambda x: x % 2 == 0, numbers))
print(result)

numbers = [5, 10, 15, 20, 25]
result = list(filter(lambda x: x > 10, numbers))
print(result)

marks = [35, 45, 55, 65, 75]
result = list(filter(lambda x: x > 50, marks))
print(result)
````

## Important concepts

Iterable, copy, exception, dictionary iteration, `pop()` / `discard()`, string formatting, type casting, dynamic typing, and `frozenset`.


### 1. Iterable

**Definition:** An iterable is an object in Python that contains multiple elements or can produce elements one at a time.

An iterable allows us to access its elements one by one using a for loop.

Examples: list, tuple, set, string, dictionary, range.


````python
numbers = [10, 20, 30, 40]
for num in numbers:
    print(num)

name = "Python"
for char in name:
    print(char)
````

### 2. Shallow copy

**Definition:** A shallow copy creates a new outer object but does not create copies of nested objects inside it.

The original object and the copied object share references to the same nested objects. Uses `copy.copy()`.


````python
import copy

original = [[10, 20], [30, 40]]
copied = copy.copy(original)
copied[0][0] = 100
print("Original:", original)
print("Copied:", copied)
````

### 3. Deep copy

**Definition:** A deep copy creates a completely independent copy of an object, including the nested objects inside it.

Changes to nested objects in the copy do not affect the original. Uses `copy.deepcopy()`.


````python
import copy

original = [[10, 20], [30, 40]]
copied = copy.deepcopy(original)
copied[0][0] = 100
print("Original:", original)
print("Copied:", copied)
````

### 4. Exception

**Definition:** An exception is an error that occurs during the execution of a Python program and interrupts its normal flow.

`try` contains code that might fail. `except` runs when an exception occurs. `else` runs when no exception occurs. `finally` always runs.


````python
a = 10
b = 0
try:
    print(a / b)
except ZeroDivisionError:
    print("Cannot divide by zero")

try:
    number = int("100")
    print(number)
except ValueError:
    print("Invalid input")
else:
    print("No exception occurred")
finally:
    print("Program completed")
````

### 5. How to iterate through a dictionary

**Definition:** Dictionary iteration means accessing keys, values or key-value pairs one by one using a loop.

- `for key in student` gives keys
- `student.values()` gives values
- `student.items()` gives key-value pairs


````python
student = {"name": "Rahul", "age": 22, "course": "Python"}

for key in student:
    print(key)

for value in student.values():
    print(value)

for key, value in student.items():
    print(key, ":", value)
````

### 6. pop() and discard()

**pop():** Removes an element and returns it. Lists, sets and dictionaries use it, with slightly different behaviour.

**discard():** Removes a specified element from a set if it exists. It does not raise an error if the element is missing, and it does not return the value.


````python
numbers = [10, 20, 30, 40]
print(numbers.pop(1))
print(numbers)

student = {"name": "Rahul", "age": 22}
print(student.pop("age"))
print(student)

nums = {10, 20, 30, 40}
nums.discard(20)
print(nums)
nums.discard(100)
print(nums)
````

**Definition:** Type casting is the process of converting a value from one data type to another data type.

`int()`, `float()`, `str()`, `bool()`, `list()`, `tuple()`, `set()`.


````python
print(int("100"), type(int("100")))
print(float(10), type(float(10)))
print(str(100), type(str(100)))
print(tuple([10, 20, 30]))
````

### 9. Dynamic typing

**Definition:** Dynamic typing means that Python determines the data type of a variable at runtime.

We do not declare the type. The same name can refer to values of different types. The type belongs to the value, not permanently to the name.


````python
x = 10
print(x, type(x))
x = "Python"
print(x, type(x))
x = 10.5
print(x, type(x))
````

### 10. frozenset

**Definition:** A frozenset is an immutable version of a set. It stores unique elements. You cannot add or remove items after creation.

A set is mutable and cannot be a dictionary key. A frozenset is immutable and can be a dictionary key if its elements are hashable.


````python
numbers = frozenset([10, 20, 30, 20])
print(numbers)

a = frozenset([1, 2, 3])
b = frozenset([3, 4, 5])
print(a.union(b))
print(a.intersection(b))
print(a.difference(b))
````

### OOP

**Definition:** Object-Oriented Programming (OOP) is a programming approach in which programs are organized around objects.

An object contains data (attributes) and actions (methods).


### Class and object

**Class:** blueprint or template used to create objects.

**Object:** instance of a class.


````python
class Student:
    pass

student1 = Student()
student2 = Student()
print(student1)
print(student2)
````

### Attributes

**Definition:** Attributes are variables that store data or properties associated with an object or class. Use the dot operator: `student1.name`.


````python
class Student:
    pass

student1 = Student()
student1.name = "Rahul"
student1.age = 15
student1.course = "Python"
print(student1.name, student1.age, student1.course)
````

### __init__() and self

**`__init__()`:** special method automatically called when a new object is created. Used to initialize attributes.

**`self`:** refers to the current object. First parameter of instance methods. Not a keyword.


````python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

s1 = Student("Rahul", 15, "Python")
s2 = Student("Priya", 17, "Python")
print(s1.name)
print(s2.name)
````

### Methods

**Definition:** A method is a function defined inside a class. It describes an action an object can perform.

Attributes store information. Methods perform actions.


````python
class Employee:
    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

    def calculate_bonus(self):
        return self.salary * 0.10

    def final_salary(self):
        return self.salary + self.calculate_bonus()

e1 = Employee("Rahul", 15000, "IT")
print(e1.calculate_bonus())
print(e1.final_salary())

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def result(self):
        return "Pass" if self.marks > 50 else "Fail"

print(Student("Rahul", 75).result())
````

### Inheritance and multiple inheritance

**Inheritance:** child class acquires attributes and methods from a parent class.

**Multiple inheritance:** one child inherits from more than one parent. Python uses Method Resolution Order (MRO).


````python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

class DataScientist(Employee):
    pass

DataScientist("Amit", 10000).display()

class Researcher:
    def research_details(self):
        print("Research details")

class DS(Employee, Researcher):
    pass

ds = DS("Amit", 10000)
ds.display()
ds.research_details()
print(DS.mro())
````

### Exception handling

**Definition:** Handle runtime exceptions so the program can respond instead of terminating unexpectedly.

`try` / `except` / optionally multiple except blocks. Use `as e` to store exception info.


````python
try:
    print(int("ABC"))
except ValueError:
    print("Please enter a valid integer.")

try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero.")

student = {"name": "Rahul", "age": 20}
try:
    print(student["salary"])
except KeyError as e:
    print("Key does not exist:", e)

try:
    print([10, 20, 30][5])
except IndexError:
    print("Index is out of range.")

try:
    print(10 + "Python")
except TypeError:
    print("Cannot add an integer and a string.")
````

### Modules

**Definition:** A module is a Python file containing reusable code such as functions, variables and classes.

`math` and `random` are standard library modules.


````python
import math
print(math.sqrt(25))
print(math.factorial(5))
print(math.ceil(4.2))
print(math.floor(4.8))

import random
print(random.randint(1, 6))
````
