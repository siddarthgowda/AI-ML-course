## Part A – Understand the concepts

### 1. What is Python? Why is it widely used?

- Python is a high-level programming language. The code reads close to plain language.
- You store a value with `name = value`. You do not write the type in front of the name.
- Python has a large set of ready-made libraries for data, automation, and AI.
- The same program can run on Windows, macOS, and Linux.

It is widely used because the syntax is short, the libraries already solve common jobs, and teams can read one another's code.

### 2. What is a programming language?

A programming language is a set of rules for writing instructions that a computer can run. Python is one programming language. Each instruction is a statement, and Python runs the statements from top to bottom.

### 3. Source code and output

Source code is the text you write. Output is the result the program produces after it runs.

Source code:

```python
print(2 + 3)
```

Output:

```text
5
```

### 4. Variables

A variable is a name attached to a value. The name is on the left of `=`. The value is on the right. You use a variable so the program can reuse that value and replace it later without rewriting the value in every line.

```python
score = 78
score = 90
```

### 5. `=` and `==`

`=` stores a value in a name. `==` checks whether two values are equal and returns `True` or `False`.

```python
a = 12
print(a == 12)
```

### 6. `int`, `float`, `str`, and `bool`

| Type | Meaning | Example |
|---|---|---|
| `int` | Whole number | `age = 21` |
| `float` | Number with a decimal point | `price = 49.50` |
| `str` | Text inside quotes | `city = "Mumbai"` |
| `bool` | `True` or `False` | `is_paid = False` |

`"123"` is a string because of the quotes. `123` is an integer.

### 7. Integer and floating-point number

An integer has no decimal point: `10`, `-3`. A floating-point number has a decimal point: `10.0`, `3.14`. Dividing with `/` produces a float, even when the numbers divide evenly: `8 / 2` is `4.0`.

### 8. What `type()` does

`type()` reports the data type of a value.

```python
print(type(10))
print(type(10.0))
print(type("10"))
print(type(True))
```

### 9. `print()` and `input()`

`print()` shows a value on the screen. `input()` pauses the program and reads what the user types. `input()` always returns a string.

### 10. Why type conversion is required for numerical input

`input()` returns text. `"10" + "20"` joins text and gives `"1020"`. A sum needs numbers, so the text is converted first: `int("10") + int("20")` gives `30`. Use `int()` for a whole number and `float()` when the value may contain a decimal point.

### 21. Arithmetic operators

| Operator | Action | Example | Result |
|---|---|---|---|
| `+` | Add | `12 + 67` | `79` |
| `-` | Subtract | `10 - 3` | `7` |
| `*` | Multiply | `4 * 5` | `20` |
| `/` | Divide | `8 / 2` | `4.0` |
| `//` | Whole-number division | `7 // 2` | `3` |
| `%` | Remainder | `7 % 2` | `1` |
| `**` | Power | `2 ** 3` | `8` |

`//` drops the decimal part. `%` keeps only the remainder. `**` raises the left value to the power of the right value.

## Part E – Challenge

### 35. Employee salary calculator

Logic:

1. Read the name, the basic monthly salary, and the bonus percentage.
2. Bonus amount = basic monthly salary × bonus percentage / 100.
3. Final monthly salary = basic monthly salary + bonus amount.
4. Annual salary = final monthly salary × 12.

Example: basic `20000` and bonus `10` give a bonus of `2000`, a final month of `22000`, and an annual salary of `264000`.


````python
# 35. Employee salary calculator.
employee_name = input("Enter the employee name: ")
basic_monthly_salary = float(input("Enter the basic monthly salary: "))
bonus_percentage = float(input("Enter the bonus percentage: "))

bonus_amount = basic_monthly_salary * bonus_percentage / 100
final_monthly_salary = basic_monthly_salary + bonus_amount
annual_salary = final_monthly_salary * 12

print("Employee:", employee_name)
print("Monthly basic salary:", basic_monthly_salary)
print("Bonus amount:", bonus_amount)
print("Final monthly salary:", final_monthly_salary)
print("Annual salary:", annual_salary)
````
