## A5. Dynamic typing

Python decides the type **at runtime**. You do not declare the type.  
**Same variable name** can point to different types later.  
The type belongs to the **value**, not permanently to the name.


```python
# Correct: same name, different types
x = 10
print(x, type(x))
x = "Python"
print(x, type(x))
x = 10.5
print(x, type(x))

def describe(value):
    print("value:", value, "| type:", type(value).__name__)

describe(1)
describe("hi")
```

## B1. if / elif / else

**if** — runs a block when the condition is `True`.  
**elif** — checks another condition if earlier ones were false.  
**else** — runs when all previous conditions were false.


```python
# Example 1: age
age = 18
if age >= 18:
    print("eligible to vote")
else:
    print("not eligible")

# Example 2: grade
grade = "B"
if grade == "A":
    print("Excellent")
elif grade == "B":
    print("Very good")
elif grade == "C":
    print("Good")
else:
    print("Invalid")

# Example 3: number ranges (like your answer)
a = 50
if a > 200:
    print("greater than 200")
elif a > 10:
    print("greater than 10")
else:
    print("invalid / small")

def check_marks(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 50:
        return "C"
    else:
        return "Fail"

print(check_marks(88), check_marks(40))
```

## B2. for loop over a list and a string

**for** walks each item of an iterable one by one.  
Use it when you already have a collection (list, string, range…).


```python
# List
numbers = [10, 20, 30, 40]
for num in numbers:
    print(num)

# String
name = "Python"
for char in name:
    print(char)

# range (also fine, but not a substitute for list+string)
for i in range(5):
    print(i)

def print_items(items):
    for item in items:
        print(item)

print_items(["Amit", "Priya"])
print_items("Hi")
```

## B3. while loop

**while** repeats as long as a condition stays `True`.  
You must update the condition (or use break), or it loops forever.

**Difference:** `for` = known collection / count. `while` = keep going until a condition fails.


```python
# Example 1: count up
n = 1
while n <= 5:
    print(n)
    n += 1

# Example 2: countdown
count = 3
while count > 0:
    print(count)
    count -= 1
print("Done")

def count_to(limit):
    i = 1
    while i <= limit:
        print(i)
        i += 1

count_to(4)
```

## B4. break vs continue

**break** — leave the loop immediately.  
**continue** — skip the rest of this round, go to the next item.


```python
# break
for i in range(1, 10):
    if i == 5:
        break
    print("break demo:", i)

# continue
for i in range(1, 8):
    if i % 2 == 0:
        continue
    print("continue demo odd:", i)

def find_first_even(nums):
    for n in nums:
        if n % 2 == 0:
            return n
    return None

print(find_first_even([1, 3, 4, 7]))
```

## C7. frozenset + shallow vs deep copy

**frozenset** = immutable set. Can be a dict key.

### C6. pop() vs discard()

**pop()** — removes an item and **returns** it.
- list: `lst.pop(index)`
- dict: `d.pop(key)`
- set: `s.pop()` removes an arbitrary item

**discard(x)** — only on a **set**. Removes `x` if present. If missing → **no error**. Does **not** return the value (`None`).

**Difference:** pop returns the removed value. discard never errors if the item is missing.


```python
# pop on list
nums = [10, 20, 30, 40]
removed = nums.pop(1)
print("Removed:", removed)
print("List:", nums)

# pop on dict
student = {"name": "Rahul", "age": 22}
val = student.pop("age")
print("Removed value:", val)
print("Dict:", student)

# discard on set
s = {10, 20, 30}
s.discard(20)
print(s)
s.discard(100)   # safe, no error
print(s)

# Wrong expectation:
print({1, 2}.discard(1))   # prints None
```

### D2. keys(), values(), items(), get()
**Difference: `d[key]` vs `d.get(key)`**

| Method | Returns |
|---|---|
| `keys()` | all keys |
| `values()` | all values |
| `items()` | key-value pairs as tuples |
| `get(key)` | value, or `None` / default if missing |

**Difference:**  
- `d[key]` → **KeyError** if key missing  
- `d.get(key, default)` → returns default, no error


```python
student = {"name": "Rahul", "age": 22, "course": "Python"}

print(student.keys())
print(student.values())
print(student.items())

print(student["name"])           # ok
print(student.get("email"))      # None
print(student.get("email", "not found"))

# student["email"]  # KeyError

def safe_get(d, key):
    return d.get(key, "missing")

print(safe_get(student, "age"))
print(safe_get(student, "phone"))
```

### D3. Loop keys, values, items

### E1. Define vs call · parameter vs argument

**Define** — write the function with `def` (logic is stored, not run yet).  
**Call** — run it with `name()`.

**Parameter** — name in the definition: `def add(a, b)`.  
**Argument** — value you pass when calling: `add(2, 3)`.


```python
# Define
def greet(name):      # name = parameter
    print("Hello", name)

# Call
greet("Amit")         # "Amit" = argument
greet("Priya")

def add(a, b):
    return a + b

print(add(10, 20))
```

### E2. print vs return

**print()** — shows a value on the screen. Function result is still `None` unless you also return.  
**return** — sends a value back to the caller so you can store or reuse it.


```python
def show_sum(a, b):
    print(a + b)          # only displays

def get_sum(a, b):
    return a + b          # gives value back

show_sum(2, 3)
result = get_sum(2, 3)
print("stored:", result)
print(result * 10)
```

### E3. Positional, keyword, default arguments

- **Positional** — matched by order: `add(2, 3)`  
- **Keyword** — matched by name: `add(b=3, a=2)`  
- **Default** — used if caller skips that argument: `def greet(name="Guest")`


```python
def intro(name, age, city="Bengaluru"):
    print(name, age, city)

intro("Rahul", 22)                      # positional + default city
intro("Priya", 25, "Mumbai")            # all positional
intro(age=20, name="Amit", city="Pune") # keyword
```

### E4. *args and **kwargs (and both)

**`*args`** — any number of positional values → stored in a **tuple**.  
**`**kwargs`** — any number of keyword values → stored in a **dictionary**.  
When combined: `def f(*args, **kwargs)` — positional first, then keyword when calling.


```python
def total(*nums):
    print(nums, type(nums))
    return sum(nums)

print(total(1, 2, 3))
print(total(10, 20))

def profile(**details):
    print(details, type(details))

profile(name="Amit", age=20, course="Python")

def both(*args, **kwargs):
    print("positional:", args)
    print("keyword:", kwargs)

both("Python", "SQL", name="Rahul", age=22)
```

### F1. List comprehension — squares + filter even

### F3. map vs filter + sorted

**map** — transform every element.

### F4. zip + enumerate + min/max

**zip** — pairs items from multiple lists (stops at shortest).  
**enumerate** — index + value while looping (`start=1` optional).

```python
names = ["Amit", "Priya", "Rahul"]
marks = [36, 67, 23]
print(list(zip(names, marks)))

for i, name in enumerate(names, start=1):
    print(i, name)

print(min(marks), max(marks))
```

### F5. Exception try / except / else / finally

**Exception** = error during execution.  
- `try` — code that might fail  
- `except` — runs if error happens  
- `else` — runs if no error  
- `finally` — always runs


```python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero")

try:
    number = int("100")
    print(number)
except ValueError:
    print("Invalid input")
else:
    print("No exception occurred")
finally:
    print("Program completed")
```

**Definition:** Object-Oriented Programming (OOP) is a programming approach in which programs are organized around objects.

An object contains data (**attributes**) and actions (**methods**).

**Why:** easier to organize, reuse code, less repetition, easier to maintain large programs.


```python
# Employee idea:
# attributes: name, age, salary, department
# methods: display details, calculate bonus
```

### G2. Class vs object

**Class:** blueprint/template used to create objects. Defines attributes and methods.

**Object:** instance of a class. Created by calling the class.

| Class | Objects |
|---|---|
| Student | Rahul, Priya |
| Car | Toyota, Honda |


```python
class Student:
    pass

student1 = Student()
student2 = Student()
print(student1)
print(student2)
```

### G3. Attributes

**Definition:** Attributes are variables that store data or properties of an object or class.

Use the **dot operator**: `student1.name`.


```python
class Student:
    pass

student1 = Student()
student1.name = "Rahul"
student1.age = 15
student1.course = "Python"
print(student1.name, student1.age, student1.course)

student2 = Student()
student2.name = "Priya"
print(student2.name)
```

### G4. __init__() and self

**`__init__()`:** special method automatically called when an object is created. Used to initialize attributes. Not a keyword.

**`self`:** refers to the current object. Usually first parameter of instance methods. Not a reserved keyword.


```python
class Student:
    def __init__(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

s1 = Student("Rahul", 15, "Python")
s2 = Student("Priya", 17, "Python")
print(s1.name)
print(s2.name)
```

### G5. Methods

**Definition:** A method is a function defined inside a class. It describes an action an object can perform.

| Attributes | Methods |
|---|---|
| Store information | Perform actions |
| name, age, salary | display(), calculate() |


```python
class Employee:
    def __init__(self, name, salary, department):
        self.name = name
        self.salary = salary
        self.department = department

    def calculate_bonus(self):
        return self.salary * 0.10

    def final_salary(self):
        return self.salary + self.calculate_bonus()

e1 = Employee("Rahul", 15000, "IT")
print(e1.calculate_bonus())
print(e1.final_salary())

class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    def result(self):
        return "Pass" if self.marks > 50 else "Fail"

print(Student("Rahul", 75).result())
```

### G6. Inheritance

**Definition:** Child class acquires attributes and methods from a parent class. Reuses code.

`class Child(Parent):`


```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Salary:", self.salary)

class DataScientist(Employee):
    pass

DataScientist("Amit", 10000).display()
```

### G7. Multiple inheritance + MRO

**Definition:** One child class inherits from more than one parent class.

**MRO:** Method Resolution Order — Python decides which parent method to use if names clash. Check with `ClassName.mro()`.


```python
class Employee:
    def employee_details(self):
        print("Employee details")

class Researcher:
    def research_details(self):
        print("Research details")

class DataScientist(Employee, Researcher):
    pass

d1 = DataScientist()
d1.employee_details()
d1.research_details()
print(DataScientist.mro())
```

### G8. Exception handling

**Definition:** Handle runtime exceptions so the program can respond instead of crashing.

`try` = code that may fail. `except` = handle the error.

| Exception | Meaning |
|---|---|
| ValueError | Correct type, invalid value |
| ZeroDivisionError | Divide by zero |
| KeyError | Missing dict key |
| IndexError | Index out of range |
| TypeError | Wrong type for operation |


```python
try:
    print(int("ABC"))
except ValueError:
    print("Please enter a valid integer.")

try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero.")

student = {"name": "Rahul", "age": 20}
try:
    print(student["salary"])
except KeyError as e:
    print("Key does not exist:", e)

try:
    print([10, 20, 30][5])
except IndexError:
    print("Index is out of range.")

try:
    print(10 + "Python")
except TypeError:
    print("Cannot add an integer and a string.")
```

### G9. Modules

**Definition:** A module is a Python file containing reusable code (functions, variables, classes).

**math** / **random** = standard library. Pandas / TensorFlow = third-party (install separately).

**Difference:** module = whole file (`math`). function = one task (`math.sqrt()`).


```python
import math
print(math.sqrt(25))
print(math.factorial(5))
print(math.ceil(4.2))
print(math.floor(4.8))

import random
print(random.randint(1, 6))
print(random.randint(1, 100))
```

- **standard vs third-party** — built-in with Python / install separately
