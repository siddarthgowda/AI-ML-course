# Part 4: File Pointer

Python remembers a position inside the open file, called the file pointer. tell() reports that position. seek() moves it, so you can read the same text again.


## Q41. What is a file pointer?

**Concept**

A file pointer represents the current position in a file
where the next read or write operation occurs.

Run the next cell. That program is the solution.


````python
# Q41 solution
from pathlib import Path

path = Path("file_demo/pointer.txt")
path.parent.mkdir(exist_ok=True)
path.write_text("PYTHON")
with path.open() as file:
    print("start", file.tell())
    print(file.read(2))
    print("after two characters", file.tell())
````

## Q42. What is tell()?

Run the next cell. That program is the solution.


````python
# Q42 solution
from pathlib import Path

path = Path("file_demo/tell_demo.txt")
path.parent.mkdir(exist_ok=True)
path.write_text("Hello")
with path.open() as file:
    file.read(2)
    print(file.tell())
````

## Q213. CUSTOM JSON ENCODER

**Concept**

JSON cannot directly serialize every Python object.

Example:

    import json
    from datetime import datetime

    data = {
        "date": datetime.now()
    }

    json.dumps(
        data,
        default=str
    )

Custom encoder:

    class MyEncoder(json.JSONEncoder):
        def default(self, obj):
            if isinstance(obj, datetime):
                return obj.isoformat()

            return super().default(obj)

Run the next cell. That program is the solution.


````python
# Q213 solution
import json
from datetime import datetime

class MyEncoder(json.JSONEncoder):
    def default(self, obj):
        if isinstance(obj, datetime):
            return obj.isoformat()
        return super().default(obj)

data = {"date": datetime(2026, 11, 5, 9, 0, 0)}
print(json.dumps(data, cls=MyEncoder))
````
